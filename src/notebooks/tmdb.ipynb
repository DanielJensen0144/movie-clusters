```python
# integrations
import sys, os, json
from dotenv import load_dotenv; load_dotenv()
print(f"workdir: {os.getcwd()}")
```

```python
import tmdbsimple as tmdb
import pandas as pd
print(f"tmdb version: {tmdb.__version__}")

tmdb.API_KEY = os.getenv("TMDB_API_KEY")
```

```python
# inspect data format
m = tmdb.Movies().popular(page=1)["results"][0]
movie_id = m["id"]
movie_details = tmdb.Movies(movie_id).info()

with open(f"./src/data/inspect/{movie_id}.json", "w") as f:
    json.dump(movie_details, f, indent=4)
    print(f"wrote movie {movie_id} to {f"./src/data/inspect/{movie_id}.json"}")

title = movie_details.get("title","")
overview = movie_details.get("overview","")
tagline = movie_details.get("tageline","")
genres = ", ".join(g["name"] for g in movie_details.get("genres",[]))
release_date = movie_details.get("release_date","")[:4]
origin_country = movie_details.get("origin_country","")
production_countries = ", ".join(c["name"] for c in movie_details.get("production_countries",[]))
production_companies = ", ".join(c["name"] for c in movie_details.get("production_companies",[]))
runtime = str(movie_details.get("runtime","")); runtime += " minutes" if runtime else ""

rich_text = (
    f"Title: {title} ({release_date}). "
    f"Genres: {genres}. "
    f"Tagline: {tagline} "
    f"Overview: {overview}. "
    f"Production companies: {production_companies}. "
    f"Origin country: {origin_country}. "
    f"Production contries: {production_countries}"
)

data = {
    "id": movie_id,
    "title": title,
    "genres": genres,
    "document": rich_text
}

with open(f"./src/data/inspect/{movie_id}-clean.json", "w") as f:
    json.dump(data, f, indent=4)
    print("wrote cleaned data to json.")
```

```python
# collect many movies
from tqdm import tqdm
n_pages = 5
movie_data = []

for page in tqdm(range(1,n_pages+1), desc="pulling pages", leave=False):
    resp = tmdb.Movies().popular(page=page)
    for m in tqdm(resp["results"], desc="pulling popular movies", leave=False):
        movie_id = m["id"]
        movie_details = tmdb.Movies(movie_id).info()

        title = movie_details.get("title","")
        overview = movie_details.get("overview","")
        tagline = movie_details.get("tageline","")
        genres = ", ".join(g["name"] for g in movie_details.get("genres",[]))
        release_date = movie_details.get("release_date","")[:4]
        origin_country = movie_details.get("origin_country","")
        production_countries = ", ".join(c["name"] for c in movie_details.get("production_countries",[]))
        production_companies = ", ".join(c["name"] for c in movie_details.get("production_companies",[]))
        runtime = str(movie_details.get("runtime","")); runtime += " minutes" if runtime else ""

        rich_text = (
            f"Title: {title} ({release_date}). "
            f"Genres: {genres}. "
            f"Tagline: {tagline} "
            f"Overview: {overview}. "
            f"Production companies: {production_companies}. "
            f"Origin country: {origin_country}. "
            f"Production contries: {production_countries}"
        )

        movie_data.append({
            "id": movie_id,
            "title": title,
            "genres": genres,
            "document": rich_text
        })

df = pd.DataFrame(movie_data)
print(f"Movie dataset: {df.shape}")
print(df.head())
```

```python
df = df.set_index("id")
df.to_csv("./src/data/inspect/top-100-popular.csv")
```
